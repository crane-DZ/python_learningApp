````python
# 第 0 节 · 自检工具箱（不用改，直接运行）
# 用法：check(lambda: 表达式, "这句是提示")
_OK, _BAD = 0, 0


def check(expr, hint=""):
    """expr 传 lambda；通过打 ✅，没写打 ⬜，写错打 ❌。返回 True/False。"""
    global _OK, _BAD
    try:
        passed = bool(expr())
    except NameError as e:
        _BAD += 1
        print(f"⬜ 还没写：{e}　→ 先去上面的动手区写代码")
        return False
    except Exception as e:
        _BAD += 1
        print(f"❌ 运行出错：{type(e).__name__}: {e}")
        return False
    if passed:
        _OK += 1
        print("✅ 通过" + (f"：{hint}" if hint else ""))
    else:
        _BAD += 1
        print("❌ 未通过" + (f"：{hint}" if hint else ""))
    return passed


def progress():
    total = _OK + _BAD
    star = "🎉 全对，可以进下一关" if total and _BAD == 0 else ""
    print(f"—— 本关进度：✅ {_OK} / {total}　{star}")


print("工具箱就绪：check() / progress() 可用")
````

## 第 3 关 · if 条件判断

📖 **书 p.136–161（第 5 章）**　🎯 第 1 周

| 写什么 | 例 |
|---|---|
| 相等 / 不等 | `temp == 25`　`temp != 25` |
| 大小 | `>=` `<=` `>` `<` |
| 组合 | `a and b`　`a or b`　`not a` |
| 在不在里面 | `x in runs`　`x not in runs` |
| 多分支 | `if / elif / else`，**只会走第一个成立的分支** |
| 空列表 | `if runs:` 为假就说明列表空 |

**判收率等级是最典型的用法**，也是第 9 关练习的骨架。注意 `elif` 的顺序：条件宽的写前面，后面的就永远轮不到。

````python
def grade(y):
    if y >= 90:
        return "优"
    elif y >= 75:
        return "良"
    elif y >= 60:
        return "合格"
    else:
        return "不合格"

def grade_bad(y):
    if y >= 60:
        return "合格"
    elif y >= 90:
        return "优"      # 永远到不了：y>=90 时上面那支已经返回了
    return "不合格"

for y in [95, 80, 62, 41]:
    print(y, "→", grade(y), "| 顺序写反的版本:", grade_bad(y))
````

````python
runs = [1, 2, 3, 5]
print("5 in runs     :", 5 in runs)
print("4 not in runs :", 4 not in runs)
print("空列表判断    :", bool([]), "| 非空:", bool(runs))
print("组合条件      :", 25 < 30 and 30 <= 35)
````

## 第 4 关 · 字典

📖 **书 p.162–189（第 6 章）**　🎯 第 1 周

字典是「一个样本的全部字段」的自然表示，也是第 9 关「练习 3 手工 groupby」的工具。

| 操作 | 写法 |
|---|---|
| 建 | `s = {"id": "S-03", "y": 62.4}` |
| 取值 | `s["y"]`（键不存在 → KeyError）｜`s.get("y", 0)`（不存在给默认值） |
| 加 / 改 | `s["note"] = "含杂质"`（键没有就是加，有就是改） |
| 删 | `del s["note"]`｜`s.pop("note", None)` |
| 键 / 值 / 键值对 | `s.keys()` `s.values()` `s.items()` |
| 长度 | `len(s)` = 键的个数 |

**`get` 和 `[]` 的区别要背下来**：`s["不存在"]` 直接崩，`s.get("不存在")` 返回 `None`。读仪器导出的字段时用 `get`，程序不会因为少一列就全挂。

````python
sample = {"id": "S-03", "temp_c": 25, "yield_pct": 62.4}
print(sample)
sample["note"] = "含杂质"
sample["yield_pct"] = 63.1
print("加/改后:", sample)

print("取值      :", sample["temp_c"])
print("get 有键  :", sample.get("temp_c"))
print("get 无键  :", sample.get("operator"))
print("get 带默认:", sample.get("operator", "未见记录"))

removed = sample.pop("note")
print("pop 返回值:", removed, "| 剩下的键:", list(sample.keys()), "| 长度:", len(sample))
````

## 第 5 关 · for 循环与列表推导式

📖 **书 p.106–122（4.1–4.3 节）、p.132–134（4.6 代码格式）**　🎯 第 2 周

这是「批量处理」的起点：一行一行洗数据、一个一个算指标，全靠它。

| 写法 | 意思 |
|---|---|
| `for x in 列表:` | 逐个取元素 |
| `for i in range(5):` | 0,1,2,3,4 |
| `for i in range(1, 11, 2):` | 1,3,5,7,9（第三步长） |
| `[表达式 for x in 列表]` | 列表推导式，一行代替三行循环 |
| `[表达式 for x in 列表 if 条件]` | 带筛选的推导式 |

**缩进就是语法**。书 p.112–115 整节都在讲缩进错误——`IndentationError` 和 `unexpected indent` 是你这两周最常见的报错，没有之一。笔记本里同一单元内缩进必须一致（4 个空格，别混 Tab）。

````python
temps_c = [25, 30, 35, 40]
temps_k = [t + 273.15 for t in temps_c]
print("摄氏度:", temps_c)
print("开尔文:", temps_k)

yields = [58.2, 61.5, 66.0, 59.9]
good = [y for y in yields if y > 60]
print(">60 的:", good, "| 个数:", len(good))

total = 0
for y in yields:
    total = total + y
print("求和（手工循环）:", round(total, 1), "| 平均:", round(total / len(yields), 2))
````

## 第 6 关 · while 循环与 input

📖 **书 p.190–209（第 7 章）**　🎯 第 2 周

`while` 用在「不知道要循环几次」的场合：等用户输入合法值、反复试直到收敛、逐行读一个大文件。

| 写法 | 意思 |
|---|---|
| `while 条件:` | 条件为真就继续 |
| `break` | 立刻跳出整个循环 |
| `continue` | 跳过本次，回到条件判断 |
| `while True:` | 必须自己用 `break` 出来，否则死循环 |

**笔记本里的坑**：`input()` 会弹输入框并**阻塞**，`Run All` 时整个笔记本会卡住。所以下面用 `fake_input()` 模拟——真实 `.py` 脚本里照书用 `input()` 就行。

**死循环急救**：如果某格一直转，点 VS Code 单元左边的 ⏹（Interrupt），别关窗口。

````python
def fake_input(prompt, value):
    """模拟 input()：把用户"输入"固定成 value，避免笔记本阻塞。"""
    print(f"{prompt}{value}  （模拟输入）")
    return value

reply = fake_input("温度（K）：", "300")
print("input 拿到的永远是字符串 →", repr(reply))
temp_k = float(reply)
print("转成浮点后才能算 →", temp_k - 273.15, "℃")

# 真实脚本里就是这两行：
# temp_k = float(input("温度（K）："))
````

````python
# while 三件套：标志、break、continue
n = 9
steps = 0
while n > 0:
    n -= 2
    steps += 1
print("9 反复减 2，减了几次?", steps, "| 剩下:", n)

total = 0
i = 0
while i < 20:
    i += 1
    if i % 3 != 0:
        continue          # 不是 3 的倍数就跳过
    total += i
print("1..20 里能被 3 整除的数的和:", total)
````

## 第 7 关 · 函数

📖 **书 p.210–247（第 8 章）**　🎯 第 2 周

函数的作用不是「代码整齐」，而是**同一件事只写错一次**。单位换算就是典型：写成一个函数，整个工程都调它，你只需要在一处保证它对。

| 写法 | 例 |
|---|---|
| 定义 | `def f(a, b): return a + b` |
| 默认值 | `def f(a, b=2): ...` |
| 关键字实参 | `f(b=3, a=1)` |
| 返回多个值 | `return mean, sd`（其实是元组） |
| 返回字典 | `return {"mean": ..., "sd": ...}`（字段多时比元组清楚） |
| 任意数量实参 | `def f(*args, **kwargs)` |

**`return` 和 `print` 不是一回事**：`print` 只是显示，`return` 才是把结果交出去。忘了 `return` 的函数返回 `None`，这是新手最隐蔽的 bug。

````python
def unit_convert(value, from_unit, to_unit):
    """支持 K↔℃、kPa↔atm、g↔mg；未知单位返回 None。"""
    if from_unit == to_unit:
        return value
    k_to_c = value - 273.15
    c_to_k = value + 273.15
    pair = (from_unit, to_unit)
    table = {
        ("K", "C"): k_to_c,
        ("C", "K"): c_to_k,
        ("atm", "kPa"): value * 101.325,
        ("kPa", "atm"): value / 101.325,
        ("g", "mg"): value * 1000,
        ("mg", "g"): value / 1000,
    }
    return table.get(pair)


print(unit_convert(300, "K", "C"))
print(unit_convert(1, "atm", "kPa"))
print(unit_convert(2, "g", "mg"))
print(unit_convert(1, "kg", "g"), "← 未知单位返回 None，不崩")
````

````python
def describe(sample_id, temp_c=25, note="正常"):
    """带默认值、返回字典。"""
    return {"id": sample_id, "temp_c": temp_c, "note": note}


d1 = describe("S-01")
d2 = describe("S-02", 30)
d3 = describe("S-03", note="含杂质")
print(d1)
print(d2)
print(d3)
````

| 写法 | 意思 |
|---|---|
| `with open(p, encoding="utf-8") as f:` | 自动关文件，**永远用 with** |
| `f.read()` / `f.readlines()` | 整个读 / 按行读成列表 |
| `"w"` / `"a"` | 覆盖写 / 追加写 |
| `pathlib.Path("data") / "x.csv"` | 拼路径，跨平台不用反斜杠 |

| `csv.reader` / `csv.DictReader` | 读 CSV，比 `split(",")` 稳 |

### 常见报错速查

| 报错 | 真实原因 | 怎么办 |
|---|---|---|
| `IndentationError` / `unexpected indent` | 缩进不一致，Tab 和空格混用 | 全删成 4 个空格 |
| `NameError: name 'x' is not defined` | 变量没定义，或本单元没运行过 | 从上面的单元依次往下跑 |
| `IndexError: list index out of range` | 索引超出长度 | `len()` 看一眼，记住从 0 开始 |
| `KeyError: 'xxx'` | 字典没这个键 | 改用 `.get("xxx", 默认值)` |
| `TypeError: can only concatenate str (not "int") to str` | 字符串和数字直接相加 | 用 f-string 或 `str()` / `int()` 转 |
| `FileNotFoundError` | 路径不对 | `Path.cwd()` 打印当前目录，确认相对路径的起点 |
| `UnicodeDecodeError` | 没写 `encoding="utf-8"` | 每个 `open` 都带上 |
